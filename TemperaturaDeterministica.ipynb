```python
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()

client = Anthropic()
model = "claude-sonnet-5-5"

system_prompt = """Eres un tutor de matemáticas paciente y claro.
- Explica paso a paso y usa ejemplos numéricos.
- Si el estudiante se equivoca, señala el error sin desanimarlo.
- Responde en español y de forma concisa.
- Si te preguntan algo que no es de matemáticas, redirige amablemente al tema."""

#La temepratura es un parámetro que controla la aleatoriedad de las respuestas del modelo.
# Entre mas baja es más consistente y predecible, mientras que una temperatura más alta 
# genera respuestas más variadas y creativas.

temperature = 0.2  # entre 0.0 (consistente) y 1.0 (variado)

messages = []

while True:
    user_input = input("> ")
    if user_input.lower() in ("salir", "exit"):
        break

    messages.append({"role": "user", "content": user_input})

    message = client.messages.create(
        model=model,
        max_tokens=1000,
        temperature=temperature,   # <- nuevo
        system=system_prompt,
        messages=messages
    )

    response = "".join(
        block.text for block in message.content if block.type == "text"
    )
    messages.append({"role": "assistant", "content": response})
    print(response)
```
